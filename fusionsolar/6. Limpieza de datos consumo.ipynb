**LIMPIEZA DE DATOS DE CONSUMOS DE LA INSTALACIÓN**
1. Determinar que datos son relevantes
2. Formatear correctamente

```python
import pandas as pd
df = pd.read_csv('inverter_consumos.csv')
df.head()
```

```python
df.describe()
```

```python
df.loc[df['xAxis'] != None, 'date_time'] = df['xAxis'].astype(str).str[:20]
df.loc[df['xAxis'] != None, 'date'] = df['xAxis'].astype(str).str[:10]
df.loc[df['xAxis'] != None, 'time'] = df['xAxis'].astype(str).str[11:16]
df.head()
```

```python
# Obteniendo el día de la semana
df.loc[df['date_time'] != None, 'weekday'] = pd.to_datetime(df['date_time'], format='mixed').dt.day_name()
df.head()
```

```python
df[df.isnull().any(axis=1)]
```

```python
df = df.dropna()
df[df.isnull().any(axis=1)]
```

```python
df.count()
```

```python
# Intenta convertir a numérico, los no numéricos se convierten a NaN
productPower_numerica = pd.to_numeric(df['productPower'], errors='coerce')
productPower_valores = ~productPower_numerica.isnull()
df = df[productPower_valores]
usePower_numerica = pd.to_numeric(df['usePower'], errors='coerce')
usePower_valores = ~usePower_numerica.isnull()
df = df[usePower_valores]
selfUsePower_numerica = pd.to_numeric(df['selfUsePower'], errors='coerce')
selfUsePower_valores = ~selfUsePower_numerica.isnull()
df = df[selfUsePower_valores]
df.head()
```

```python
# Ajustando a valores anómalos
df.rename(columns={'productPower': 'generado'}, inplace=True)
df.rename(columns={'usePower': 'consumido'}, inplace=True)
df.rename(columns={'selfUsePower': 'autoconsumo'}, inplace=True)
df.head()
```

```python
# Nos quedamos con las columnas que nos interesan
df_ok = df[['date_time', 'date', 'time', 'weekday', 'generado', 'consumido', 'autoconsumo']]
```

```python
df_ok.to_csv('inverter_consumos_clean.csv', index=False)
```
