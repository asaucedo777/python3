# 1. Introducción
La instalación de las placas solares en la vivienda ha tenido beneficios significativos en el ahorro energético desde su instalación en septiembre de 2022. 
No obstante en los últimos meses hemos notado un aumento de la factura de la luz. Por este motivo hemos considerado interesante analizar las causas de dicho incremento y más concretamente el estado de la instalación de las placas y su producción eléctrica.

# 2. Objetivo
El objetivo principal del presente estudio es determinar el estado actual de la instalación para decidir si realizar alguna revisión técnica y la posible ampliación de la instalación para mejorar el ahorro energético.


# 3. Indicadores
Los indicadores que nos intereses son:
- Producción eléctrica máxima
- Producción eléctrica media
- Autoconsumo
- Análisis estacional mensual
- Análisis estacional semanal
- Análisis estacional diario

# 4. Importación de datos
Las fuentes de información para el análisis son las facturas eléctricas desde septiembre de 2022 a febrero de 2025, así como la información importada del inversor de las placas, que recoge información horaria desde el día 2024-02-17 04:00:00.
* La información importada del inversor se encuentra en el archivo "inverter_horario.json"

```python
import pandas as pd
df = pd.read_json('inverter_horario.json')
df.tail(1)
```

```python
df_plain = pd.json_normalize(df['data'])
df_plain.head(1)
```

```python
list_with_dict = df_plain['list'].apply(lambda x: [dict(item) for item in x])
list_with_dict.count()
```

```python
list_exploded = list_with_dict.explode()
list_exploded.head()
```

```python
df_final = pd.json_normalize(list_exploded)
df_final.head(1)
```

```python
df_final.to_csv('inverter_horario.csv', index=False)
```
