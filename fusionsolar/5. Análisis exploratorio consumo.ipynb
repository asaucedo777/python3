# 1. Introducción
En los siguientes notebooks voy a analizar los datos de consumo elétrico recogidos por el inversor. Los datos que se han podido recoger de la web de huawei son desde el día 12 de agosto de 2023 y recogen lecturas cada 5 minutos.

# 2. Objetivo
El objetivo principal del presente estudio es describir el consumo durante los casi 2 años de datos. Concretamente nos interesa el comportamiento estacional del consumo, así como las anomalías en el mismo, de cara a tomar decisiones relatias a la modificación de la instalación, la revisión técnica o la reclamación de lecturas incorrectas por parte de la compañía eléctrica.


# 3. Indicadores
Los indicadores que nos intereses son:
- Consumo mínimo y máximo durante todo oel periodo para adapatar la potencia contratada
- Consumo mínimo y máximo diario
- Serie de consumo medio diario

# 4. Importación de datos
* La información importada del inversor se encuentra en el archivo "inverter_consumo.json"

```python
import pandas as pd
df = pd.DataFrame(pd.read_json('inverter_consumos.json')['data'])
df.head(1)
```

```python
df_plain = pd.json_normalize(df['data'])
df_plain.head(1)
```

```python
df_target = df_plain[['productPower', 'usePower', 'selfUsePower', 'xAxis']]
df_target.head(2)
```

```python
df_exploded = df_target.apply(pd.Series.explode)
df_exploded.head(2)
```

```python
df_exploded.dropna()
df_exploded.count()
```
