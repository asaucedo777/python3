**LIMPIEZA DE DATOS DE LECTURA DEL INVERSOR**
1. Determinar que datos son relevantes
2. Formatear correctamente

```python
import pandas as pd
df = pd.read_csv('inverter_horario.csv')
df.head()
```

```python
df.describe()
```

```python
df['inverterPower'] == df['productPower']
```

### Descartamos las columnas powerProfit (por no existir datos relevantes) y productPower (por ser equivalente a inverterPower)

```python
df_clean = df[['fmtCollectTimeStr', 'inverterPower']]
df_clean.head()
```

```python
df_clean.loc[df_clean['fmtCollectTimeStr'] != None, 'date_time'] = df_clean['fmtCollectTimeStr'].astype(str).str[:20]
df_clean.loc[df_clean['fmtCollectTimeStr'] != None, 'date'] = df_clean['fmtCollectTimeStr'].astype(str).str[:10]
df_clean.loc[df_clean['fmtCollectTimeStr'] != None, 'time'] = df_clean['fmtCollectTimeStr'].astype(str).str[11:16]
df_clean.head()
```

```python
# Obteniendo el día de la semana
df_clean.loc[df_clean['date_time'] != None, 'weekday'] = pd.to_datetime(df_clean['date_time'], format='mixed').dt.day_name()
df_clean.head()
```

```python
# Redondeando a dos decimales
df_clean['inverterPower'] = df_clean['inverterPower'].round(2)
df_clean.head()
```

```python
# Nos quedamos con las columnas que nos interesan
df_clean_ok = df_clean[['date_time', 'date', 'time', 'weekday', 'inverterPower']]
```

```python
df_clean_ok.to_csv('inverter_horario_clean.csv', index=False)
```
