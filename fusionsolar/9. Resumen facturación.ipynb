**RESUMEN FACTURACION AÑOS 2022 A 2025**

```python
import pandas as pd
df = pd.read_json('consumos_facturados.json')
df.head()
```

```python
df_graficoConsumos = pd.json_normalize(df['graficoConsumos'])
df_graficoConsumos.columns = ['graficoConsumos.' + col for col in df_graficoConsumos.columns]
df_graficoConsumos.head()
```

```python
df_graficoImportes = pd.json_normalize(df['graficoImportes'])
df_graficoImportes.columns = ['graficoImportes.' + col for col in df_graficoImportes.columns]
df = pd.concat([df, df_graficoConsumos, df_graficoImportes], axis=1)
df.head()
```

```python
df = df[[
  'graficoConsumos.pie',
  'graficoConsumos.fecFactura', 
  'graficoConsumos.data',
  'graficoConsumos.serieConsumo',
  'graficoImportes.pie',
  'graficoImportes.fecMaxTerEne', 
  'graficoImportes.fecMinTerEne',
  'graficoImportes.fecMinTerPot', 
  'graficoImportes.fecMaxTerPot',
  'graficoImportes.fecMaxTer', 
  'graficoImportes.fecMinTer',
  'graficoImportes.numFactura', 
  'graficoImportes.importeFactura',
  'graficoImportes.data', 
  ]]
df.head()
```

```python
df_serieConsumo = pd.json_normalize(df['graficoConsumos.serieConsumo'])
df_serieConsumo.head()
```

```python
df = df.drop(columns=['graficoConsumos.serieConsumo'])
df.head()
```

```python
df.columns
```

```python
series_dict = {col: df[col].explode().reset_index(drop=True) for col in df.columns}
series_dict['graficoConsumos.pie']
```

```python
df_clean = pd.DataFrame(series_dict)
df_clean.head()
```

```python
df_clean = df_clean.drop_duplicates()
df_clean.count()
```

```python
df_clean['graficoConsumos.anno_mes'] = pd.to_datetime(df_clean['graficoConsumos.pie'], format='%d/%m/%Y').dt.strftime('%y-%m')
```

```python
# Renombra las columnas
df_clean.rename(columns={
  'graficoConsumos.anno_mes': 'Año-mes',
  'graficoConsumos.pie': 'Fecha consumo',
  'graficoConsumos.fecFactura': 'Fecha Factura',
  'graficoConsumos.data': 'Consumo',
  'graficoImportes.pie': 'Fecha Facturación',
  'graficoImportes.fecMaxTerEne': 'Fecha Maxima Término Energia',
  'graficoImportes.fecMinTerEne': 'Fecha Minima Término Energia',
  'graficoImportes.fecMinTerPot': 'Fecha Minima Término Potencia',
  'graficoImportes.fecMaxTerPot': 'Fecha Maxima Término Potencia',
  'graficoImportes.fecMaxTer': 'Fecha Máxima Término',
  'graficoImportes.fecMinTer': 'Fecha Minima Término',
  'graficoImportes.numFactura': 'Numero Factura',
  'graficoImportes.importeFactura': 'Importe Factura',
  'graficoImportes.data': 'Importe',
}, inplace=True)
```

```python
# Ajustar formato de las fechas yyyy-mm-dd
import locale
locale.setlocale(locale.LC_TIME, 'es_ES.UTF-8')
df_clean['Fecha consumo'] = pd.to_datetime(df_clean['Fecha consumo'], format='%d/%m/%Y').dt.strftime('%Y-%m-%d')
df_clean['Fecha Factura'] = pd.to_datetime(df_clean['Fecha Factura'], format='%d/%m/%Y').dt.strftime('%Y-%m-%d')
df_clean['Fecha Facturación'] = pd.to_datetime(df_clean['Fecha Facturación'], format='%d/%m/%Y').dt.strftime('%Y-%m-%d')
df_clean['Fecha Maxima Término Energia'] = pd.to_datetime(df_clean['Fecha Maxima Término Energia'], format='%d de %B de %Y').dt.strftime('%Y-%m-%d')
df_clean['Fecha Minima Término Energia'] = pd.to_datetime(df_clean['Fecha Minima Término Energia'], format='%d de %B de %Y').dt.strftime('%Y-%m-%d')
df_clean['Fecha Minima Término Potencia'] = pd.to_datetime(df_clean['Fecha Minima Término Potencia'], format='%d de %B de %Y').dt.strftime('%Y-%m-%d')
df_clean['Fecha Maxima Término Potencia'] = pd.to_datetime(df_clean['Fecha Maxima Término Potencia'], format='%d de %B de %Y').dt.strftime('%Y-%m-%d')
df_clean['Fecha Máxima Término'] = pd.to_datetime(df_clean['Fecha Máxima Término'], format='%d de %B de %Y').dt.strftime('%Y-%m-%d')
df_clean['Fecha Minima Término'] = pd.to_datetime(df_clean['Fecha Minima Término'], format='%d de %B de %Y').dt.strftime('%Y-%m-%d')
```

```python
df_clean.to_csv('consumos_facturados_clean.csv', index=False)
```
