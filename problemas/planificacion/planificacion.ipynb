### Un robot se tiene que desplazar por una cuadrícula desde el punto A al punto B, utilizando sólo movimientos horizontales o verticales

```python
import networkx as nx
import matplotlib.pyplot as plt

print("Creating a graph")
```

```python
def create_graph(n):
    grafo = nx.Graph()
    for i in range(n):
        for j in range(n):
            nodo = (i, j)
            grafo.add_node(nodo)
            # Conexión con el vecino de arriba
            if i > 0:
                grafo.add_edge(nodo, (i - 1, j))
            # Conexión con el vecino de abajo
            if i < n - 1:
                grafo.add_edge(nodo, (i + 1, j))
            # Conexión con el vecino de la izquierda
            if j > 0:
                grafo.add_edge(nodo, (i, j - 1))
            # Conexión con el vecino de la derecha
            if j < n - 1:
                grafo.add_edge(nodo, (i, j + 1))
    return grafo
def heuristicManhattan(node1, node2):
    return abs(node1[0] - node2[0]) + abs(node1[1] - node2[1])
def pathAstar(graph, ini, end):
    return nx.astar_path(graph, ini, end, heuristic=heuristicManhattan)
def main():
    N = 10
    ini = (0, 0)
    end = (4, 6)
    graph = create_graph(N)
    path = pathAstar(graph, ini, end)
    print(path)
    
    pos = {(x, y): (x, y) for x in range(N) for y in range(N)}
    
    nx.draw(graph, pos, with_labels=True)
    nx.draw_networkx_nodes(graph, pos, nodelist=[ini], node_color='g')
    nx.draw_networkx_nodes(graph, pos, nodelist=[end], node_color='e')
    nx.draw_networkx_nodes(graph, pos, ed)
    plt.show()
```
