```python
import heapq

def dijkstra(grafo, inicio):
  # Inicializamos las distancias de los nodos a INFINITO
  distancias = {nodo: float('infinity') for nodo in grafo}
  # La distancia del nodo de inicio a si mismo es 0
  distancias[inicio] = 0
  # Definimos una lista de nodos a visitar
  cola_de_prioridad = []
  # Inicializamos la cola de prioridad con el nodo de inicio
  heapq.heappush(cola_de_prioridad, (0, inicio))
  
  while cola_de_prioridad:
    distancia_actual, nodo_actual = heapq.heappop(cola_de_prioridad)
    # Si la distancia actual es mayor a la distancia almacenada
    # en el diccionario, no hacemos nada
    if distancia_actual > distancias[nodo_actual]:
      continue
    # Iteramos sobre los vecinos del nodo actual
    for vecino, peso in grafo[nodo_actual].items():
      distancia = distancia_actual + peso
      # Si la distancia es menor a la almacenada en el diccionario
      # actualizamos la distancia y agregamos el nodo a la cola
      if distancia < distancias[vecino]:
        distancias[vecino] = distancia
        heapq.heappush(cola_de_prioridad, (distancia, vecino))
  return distancias

# Definimos un grafo con pesos
grafo = {
    'Madrid': {'Barcelona': 621, 'Sevilla': 528, 'Toledo': 72},
    'Barcelona': {'Madrid': 621, 'Sevilla': 996, 'Toledo': 683},
    'Sevilla': {'Madrid': 528, 'Toledo': 589},
    'Toledo': {'Madrid': 72, 'Barcelona': 683, 'Sevilla': 589}
}

inicio = 'Sevilla'
print(dijkstra(grafo, inicio))
```
